```python
import smtplib
import pandas as pd
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText

# Step 1: Read the uploaded Excel file into a DataFrame
file_name = "A2_marks.xlsx"  # replace with your file name if different
df = pd.read_excel(file_name)
count = 0
# Step 2: Set up your SMTP server (Using Gmail SMTP server here)
SMTP_SERVER = 'smtp.gmail.com'
SMTP_PORT = 587

# Replace with your own Gmail credentials
SENDER_EMAIL = 'test@email.com'  # Your Gmail address
SENDER_PASSWORD = 'aaaa bbbb cccc dddd'  # Your Gmail password or App Password

# Step 3: Set up the email subject and message template
subject = "2402 SSE Introductory Biology (BIO  101 S1) : Assignment 2 Announcement"

# Step 4: Set up the SMTP connection
server = smtplib.SMTP(SMTP_SERVER, SMTP_PORT)
server.starttls()  # Upgrade the connection to secure TLS mode
server.login(SENDER_EMAIL, SENDER_PASSWORD)  # Login with your email and password

# Step 5: Loop through the DataFrame and send emails
for index, row in df.iterrows():
    name = row['Name']
    email = row['Email ID ']
    marks = row['A2 Marks']
    # variant = row['Variant']
    # marks_out_of5_5 = row['Marks(out of5.5%)']

    # Create the message object
    msg = MIMEMultipart()
    msg['From'] = SENDER_EMAIL
    msg['To'] = email
    msg['Subject'] = subject

    # Create the body of the email
    body = f"""
Dear {name},

Hoping that you are doing well.

Following are your marks for Assignment 2:

Your score is: {marks}/20
The mean for A2 is: 18.84
The standard deviation is: 1.69


Best regards,
Bio101 Teaching Assistants Team


*Note: This is an automated email. Please do not reply. Feel free to reach out to TAs for further clarification*

    """

    # Attach the body to the email
    msg.attach(MIMEText(body, 'plain'))

    # Send the email
    try:
        server.sendmail(SENDER_EMAIL, email, msg.as_string())
        print(f"Email sent successfully to {email}")
        count = count + 1
    except Exception as e:
        print(f"Failed to send email to {email}. Error: {e}")

print(count)

# Step 6: Quit the SMTP server after sending all emails
server.quit()
```
